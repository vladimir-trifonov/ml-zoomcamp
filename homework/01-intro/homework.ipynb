```python
car_data = pd.read_csv("car_fuel_efficiency_2026.csv")
print(car_data.head())
```

```python
count = len(car_data)
print(f"Number of rows in the dataset: {count}")
```

```python
fuel_types = len(car_data['fuel_type'].unique())
print(f"Unique fuel types in the dataset: {fuel_types}")
```

```python
missing_values = car_data.isnull().sum()
print("Missing values in each column:")
print(len(missing_values[missing_values > 0]))
```

```python
max_fuel_efficiency = car_data[car_data['origin'] == 'Asia']['fuel_efficiency_mpg'].max()
print("Car(s) with the highest fuel efficiency:")
print(max_fuel_efficiency)
```

```python
median_horsepower = car_data['horsepower'].median()
print(f"Median horsepower of all cars: {median_horsepower}")
most_frequent_horsepower = car_data['horsepower'].mode()[0]
print(f"Most frequent horsepower value: {most_frequent_horsepower}")
car_data.fillna({'horsepower': most_frequent_horsepower}, inplace=True)
new_median_horsepower = car_data['horsepower'].median()
print(f"New median horsepower after filling missing values: {new_median_horsepower}")
```

```python
asia_cars = car_data[car_data['origin'] == 'Asia'][['vehicle_weight', 'model_year']]
print("Vehicle weight and model year of cars from Asia:")
print(asia_cars[:7])
X = asia_cars[:7].to_numpy()
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv.dot(X.T).dot(y)
print("Weights for the linear regression model:")
print(sum(w))
```
